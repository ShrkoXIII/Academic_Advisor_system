```python
id_cols_to_check = [
    "student_status_id",
    "student_id",
    "part_id",
    "degree_id",
    "start_part_id",
    "finish_part_id",
    "start_level_id",
]

for col in id_cols_to_check:
    if col in df.columns:
        s = df[col].dropna().astype(str)

        # الجزء بعد النقطة
        suffix = s.str.extract(r"\.(.*)$")[0]

        print("\nCOLUMN:", col)
        print("non-null:", df[col].notna().sum())
        print("null:", df[col].isna().sum())
        print("sample values:")
        print(df[col].dropna().head(10).tolist())
        print("top suffixes:")
        print(suffix.value_counts(dropna=False).head(10))
```

```python
course_count_cols = [
    "semester_reg_courses",
    "semester_pass_courses",
    "semester_fail_courses",
    "total_pass_courses",
    "total_fail_courses",
    "reg_total_semesters",
]

for col in course_count_cols:
    if col in df.columns:
        s = pd.to_numeric(df[col], errors="coerce")
        non_null = s.dropna()
        fractional = non_null[non_null % 1 != 0]

        print("\nCOLUMN:", col)
        print("non-null:", non_null.shape[0])
        print("null:", s.isna().sum())
        print("fractional count:", fractional.shape[0])
        if fractional.shape[0] > 0:
            print("fractional examples:", fractional.head(10).tolist())
```

```python
print("Rows:", len(df))
print("Unique student_status_id:", df["student_status_id"].nunique(dropna=True))
print("Duplicated student_status_id rows:", df.duplicated(subset=["student_status_id"], keep=False).sum())

display(
    df[df.duplicated(subset=["student_status_id"], keep=False)]
    .sort_values("student_status_id")
    .head(20)
)
```

```python
gpa_agpa_cols = [
    "gpa_percent",
    "gpa_points",
    "start_agpa_percent",
    "start_agpa_points",
    "end_agpa_percent",
    "end_agpa_points",
]

missing_gpa_agpa_rows = df[df[gpa_agpa_cols].isna().any(axis=1)]

print("Rows with missing GPA/AGPA:", len(missing_gpa_agpa_rows))

display(
    missing_gpa_agpa_rows[
        [
            "student_status_id",
            "student_id",
            "part_id",
            "degree_id",
            "study_mode",
            "gpa_percent",
            "gpa_points",
            "start_agpa_percent",
            "start_agpa_points",
            "end_agpa_percent",
            "end_agpa_points",
            "semester_reg_courses",
            "semester_reg_credits",
            "version_title_sl",
        ]
    ]
)
```
